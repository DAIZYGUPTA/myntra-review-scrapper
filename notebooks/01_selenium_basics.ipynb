```python

from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

import time
```

```python
driver = webdriver.Chrome(
    service=Service(ChromeDriverManager().install())
)

driver.maximize_window()
url = "https://www.myntra.com/cotton-tshirt?rawQuery=cotton%20tshirt"

driver.get(url)
```

```python
wait = WebDriverWait(driver, 15)

title = wait.until(
    EC.presence_of_element_located(
        (By.TAG_NAME, "h1")
    )
)

print(title.text)
```

```python
products = driver.find_elements(
    By.CSS_SELECTOR,
    "li.product-base"
)

print(len(products))
```

```python
first_product = products[0]
```

```python
brand = first_product.find_element(
    By.CSS_SELECTOR,
    "h3.product-brand"
)

print(brand.text)
```

```python
product_name = first_product.find_element(
    By.CSS_SELECTOR,
    "h4.product-product"
)

print(product_name.text)
```

```python
price = first_product.find_element(
    By.CSS_SELECTOR,
    "span.product-discountedPrice"
)

print(price.text)
```

```python
rating = first_product.find_element(
    By.CSS_SELECTOR,
    "div.product-ratingsContainer span"
)

print(rating.text)
```

```python
link = first_product.find_element(
    By.TAG_NAME,
    "a"
)

print(link.get_attribute("href"))
```

```python
image = first_product.find_element(
    By.TAG_NAME,
    "img"
)

print(image.get_attribute("src"))
```

```python
product_data = {
    "brand": brand.text,
    "product_name": product_name.text,
    "price": price.text,
    "rating": rating.text,
    "product_url": link.get_attribute("href"),
    "image_url": image.get_attribute("src")
}

print(product_data)
```

```python
all_products = []

products = driver.find_elements(
    By.CSS_SELECTOR,
    "li.product-base"
)

for product in products:

    brand = product.find_element(
        By.CSS_SELECTOR,
        "h3.product-brand"
    ).text

    product_name = product.find_element(
        By.CSS_SELECTOR,
        "h4.product-product"
    ).text

    price = first_product.find_element( By.CSS_SELECTOR, "span.product-discountedPrice")
    #price = product.find_element(By.CSS_SELECTOR,"span.product-discountedPrice" ).text

    data = {
        "brand": brand,
        "product_name": product_name,
        "price": price
    }

    all_products.append(data)

print(len(all_products))
print(all_products[:3])
```
