```python
# Create Scroll Function

def scroll_to_bottom(driver):

    last_height = driver.execute_script(
        "return document.body.scrollHeight"
    )

    while True:

        driver.execute_script(
            "window.scrollTo(0, document.body.scrollHeight);"
        )

        time.sleep(3)

        new_height = driver.execute_script(
            "return document.body.scrollHeight"
        )

        if new_height == last_height:
            break

        last_height = new_height
```

```python
# extract product URLs

def get_product_links(driver):

    product_links = []

    products = driver.find_elements(
        By.CSS_SELECTOR,
        "li.product-base"
    )

    for product in products:

        try:

            link = product.find_element(
                By.TAG_NAME,
                "a"
            ).get_attribute("href")

            product_links.append(link)

        except:
            pass

    return product_links
```

```python
# Open Category Page

url = "https://www.myntra.com/cotton-tshirts?rawQuery=cotton%20tshirts"

driver.get(url)
```

```python
scroll_to_bottom(driver)
```

```python
# collect product Links

product_links = get_product_links(driver)

print(len(product_links))
```

```python
def extract_reviews(driver):

    all_reviews = []

    reviews = driver.find_elements(
        By.CSS_SELECTOR,
        "div.user-review-userReviewWrapper"
    )

    for review in reviews:

        try:

            rating = review.find_element(
                By.CSS_SELECTOR,
                "span.user-review-starRating"
            ).text

            review_text = review.find_element(
                By.CSS_SELECTOR,
                "div.user-review-reviewTextWrapper"
            ).text

            username = review.find_element(
                By.CSS_SELECTOR,
                "div.user-review-left span:nth-child(1)"
            ).text

            date = review.find_element(
                By.CSS_SELECTOR,
                "div.user-review-left span:nth-child(2)"
            ).text

            review_data = {
                "rating": rating,
                "review_text": review_text,
                "username": username,
                "date": date
            }

            all_reviews.append(review_data)

        except Exception as e:
            print(e)

    return all_reviews
```

```python
# Build Product Visit Loop

all_data = []

for index, product_url in enumerate(product_links):

    try:

        print(f"Opening Product {index+1}")

        driver.get(product_url)

        time.sleep(5)

        # Scroll slightly to review area
        driver.execute_script(
            "window.scrollTo(0, document.body.scrollHeight/2);"
        )

        time.sleep(3)

        reviews = extract_reviews(driver)

        for review in reviews:

            review["product_url"] = product_url

            all_data.append(review)

    except Exception as e:

        print("Product Error:", e)
```
