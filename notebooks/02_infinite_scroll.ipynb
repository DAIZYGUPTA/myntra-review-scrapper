```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager

from selenium.webdriver.common.by import By

import pandas as pd
import time
```

```python
driver = webdriver.Chrome(
    service=Service(ChromeDriverManager().install())
)

driver.maximize_window()

url = "https://www.myntra.com/cotton-tshirt?rawQuery=cotton%20tshirt"

driver.get(url)
```

```python
height = driver.execute_script(
    "return document.body.scrollHeight"
)

print(height)
```

```python
last_height = driver.execute_script(
    "return document.body.scrollHeight"
)

while True:

    # Scroll down
    driver.execute_script(
        "window.scrollTo(0, document.body.scrollHeight);"
    )

    # Wait for loading
    time.sleep(3)

    # New page height
    new_height = driver.execute_script(
        "return document.body.scrollHeight"
    )

    print("Old Height:", last_height)
    print("New Height:", new_height)

    # Stop if no new content
    if new_height == last_height:
        print("Reached end of page")
        break

    last_height = new_height
```

```python
# Create Scroll Function

def scroll_to_bottom(driver):

    last_height = driver.execute_script(
        "return document.body.scrollHeight"
    )

    while True:

        driver.execute_script(
            "window.scrollTo(0, document.body.scrollHeight);"
        )

        time.sleep(3)

        new_height = driver.execute_script(
            "return document.body.scrollHeight"
        )

        if new_height == last_height:
            break

        last_height = new_height
```

```python
# Create Product Extraction Function

def extract_products(driver):

    all_products = []

    products = driver.find_elements(
        By.CSS_SELECTOR,
        "li.product-base"
    )

    for product in products:

        try:

            brand = product.find_element(
                By.CSS_SELECTOR,
                "h3.product-brand"
            ).text

            product_name = product.find_element(
                By.CSS_SELECTOR,
                "h4.product-product"
            ).text

            price = product.find_element(
                By.CSS_SELECTOR,
                "span.product-discountedPrice"
            ).text

            data = {
                "brand": brand,
                "product_name": product_name,
                "price": price
            }

            all_products.append(data)

        except Exception as e:
            print(e)

    return all_products
```

```python
# Main Pagination Loop

all_data = []

page = 1

while True:

    print(f"Scraping Page {page}")

    # Scroll current page fully
    scroll_to_bottom(driver)

    # Extract products
    products = extract_products(driver)

    all_data.extend(products)

    try:

        next_button = driver.find_element(
            By.CSS_SELECTOR,
            "li.pagination-next"
        )

        driver.execute_script(
            "arguments[0].click();",
            next_button
        )

        time.sleep(5)

        page += 1

    except:
        print("No more pages")
        break
```
